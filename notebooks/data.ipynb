```python
ex = ds[0]
print("QUESTION:", ex["question"])
print()
print("ANSWER:", ex["final_decision"])
print()
print("LONG ANSWER:", ex["long_answer"])
print()
for label, text in zip(ex["context"]["labels"], ex["context"]["contexts"]):
    print(f"[{label}] {text}\n")
```

```python
art = load_dataset("qiaojin/PubMedQA", "pqa_artificial", split="train")
print(art)
```

```python
corpus = []

for ex in ds:
    doc = {
        "doc_id": str(ex["pubid"]),
        "text": " ".join(ex["context"]["contexts"]),
    }
    corpus.append(doc)

print(len(corpus))
```

```python
for ex in art.select(range(5000)):
    doc = {
        "doc_id": str(ex["pubid"]),
        "text": " ".join(ex["context"]["contexts"]),
    }
    corpus.append(doc)

print(len(corpus))
```

```python
ids = [doc["doc_id"] for doc in corpus]
print(len(ids), len(set(ids)))
```

```python
questions = []
for ex in ds:
    questions.append({
        "question": ex["question"],
        "label": ex["final_decision"],
        "gold_doc_id": str(ex["pubid"]),
    })

print(len(questions))
print(questions[0])
```

```python
import json

with open("../data/corpus.json", "w") as f:
    json.dump(corpus, f)

with open("../data/questions.json", "w") as f:
    json.dump(questions, f)
```
